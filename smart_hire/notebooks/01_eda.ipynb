# 01 - Exploratory Data Analysis

Inspect the supplied resume and job datasets before modeling.

```python
from pathlib import Path
import json
import pandas as pd

PROJECT_ROOT = Path.cwd()
while PROJECT_ROOT.name != "smart_hire" and PROJECT_ROOT != PROJECT_ROOT.parent:
    PROJECT_ROOT = PROJECT_ROOT.parent
DATA_ROOT = PROJECT_ROOT.parent
jobs_path = next(DATA_ROOT.glob("marketing_sample_for_naukri_com-naukri_com_job_data__20201001_20201231__5k_data.csv/*.ldjson"))
resume_path = DATA_ROOT / "Resume_cleaned.csv"

with jobs_path.open(encoding="utf-8") as handle:
    jobs = pd.DataFrame(json.loads(line) for line in handle if line.strip())
resumes = pd.read_csv(resume_path)
print(f"Jobs: {jobs.shape[0]:,} rows, {jobs.shape[1]} columns")
print(f"Resumes: {resumes.shape[0]:,} rows, {resumes.shape[1]} columns")
```

```python
print("Top job categories:")
print(jobs["category"].value_counts().head(10))
print("\nMost common job cities:")
print(jobs["city"].astype(str).str.split("|").explode().value_counts().head(10))

summary = jobs[["job_title", "category", "company_name", "city"]].head(10)
summary
```
