# 04 - Clustering and Topics

Group job descriptions with KMeans and inspect the strongest terms in each cluster.

```python
from pathlib import Path
import json
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.feature_extraction.text import TfidfVectorizer

PROJECT_ROOT = Path.cwd()
while PROJECT_ROOT.name != "smart_hire" and PROJECT_ROOT != PROJECT_ROOT.parent:
    PROJECT_ROOT = PROJECT_ROOT.parent
DATA_ROOT = PROJECT_ROOT.parent
jobs_path = next(DATA_ROOT.glob("marketing_sample_for_naukri_com-naukri_com_job_data__20201001_20201231__5k_data.csv/*.ldjson"))
with jobs_path.open(encoding="utf-8") as handle:
    jobs = pd.DataFrame(json.loads(line) for line in handle if line.strip()).fillna("")
jobs = jobs.head(500).copy()
texts = (jobs["job_title"] + " " + jobs["job_description"]).astype(str)
vectorizer = TfidfVectorizer(stop_words="english", max_features=2500, min_df=2)
matrix = vectorizer.fit_transform(texts)
print(matrix.shape)
```

```python
cluster_count = 6
model = KMeans(n_clusters=cluster_count, random_state=42, n_init=10)
jobs["cluster"] = model.fit_predict(matrix)
terms = vectorizer.get_feature_names_out()
for cluster_id, center in enumerate(model.cluster_centers_):
    top_terms = terms[center.argsort()[-10:][::-1]]
    print(f"Cluster {cluster_id}: {', '.join(top_terms)}")

jobs[["job_title", "cluster"]].head(10)
```
